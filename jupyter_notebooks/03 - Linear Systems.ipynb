## Standard Fundamental System

From any fundamental matrix $\tn V(t)$ every solution can be written as:

$$
    \vc v(t) = \tn V(t) \vc c 
$$

for some vector $\vc c$ in $\Real^n$, however there is the unique *standard fundamental matrix* (SFM) $\tn U(t, \tau)$ for which
this coefficient vector is just the initial condition of the solution, i.e. $\vc c = \vc v(\tau) = \vc \xi$.

To find the standard fundamental matix, let any fundamental matrix $\tn V(t)$ be given.
To find $\vc c$ for given $\vc \xi$ we calculate:

$$
\vc c = \tn V(\tau)^{-1} \vc \xi,\quad \vc v(t) = \Big[ \tn V(t) \tn V(\tau)^{-1} \Big] \xi
$$

So the standard fundamental matrix can be obtained as:

$$
    \tn U(t, \tau) = \tn V(t) \tn V^{-1}(\tau)
$$

for any fundamental matrix $\tn V(t)$. Th columns of SFM form the standard fundamental system, that is the basis of the solution space formed by solutions $\vc x(t; \tau, \vc e_i)$ with the base vectors $\vc e_i$ of $\Real^n$ as the initial conditions.

We also call $\tn U(t, \tau)$ the *transition matrix* as its multiplication realizes transition of a solution from the initial time $\tau$ to other time $t$. The standard fundamental matrix has all properties of the general fundamental matrices, but have a few interesting properties on its own:

**Theorem**
Let ${\tn U}(t,\tau)$ be the standard fundamental matrix of the homogeneous system:

$$
    \dot{\vc x} = {\tn A}(t)\vc x,
$$

for $t, \tau\in I$. Then:

1. ${\tn U}(\tau,\tau)$ is the identity matrix for all $\tau\in I$.
1. ${\tn U}(t,\tau){\tn U}(\tau,s)={\tn U}(t,s)$ for all $t, \tau, s\in I$ .
1. $[{\tn U}(t,\tau)]^{-1}={\tn U}(\tau,t)$ for all $t, \tau\in I$.

**Proof:**
1. Column $i$ of the matrix $\tn U(\tau, \tau)$ is the solution $\vc x( \cdot; \tau, \vc e_i)$ in time $\tau$ where it satisfies the initial condition:
$$
    \vc x(\tau; \tau, \vc e_i) = \vc e_i.
$$
This vector has one only on row $i$, i.e. the matrix $\tn U(\tau, \tau)$ has ones only on the diagonal, zeros elsewhere.
 
1. For any $\vc \xi \in \Real^n$ we have
$$
    \tn U(t,s) \vc \xi =
    \vc x(t;s, \xi) = \vc x(t; \tau, \vc x(\tau; s, \vc \xi)) =
    \tn U(t,\tau) \vc x(\tau; s, \vc \xi) =
    \tn U(t, \tau) \tn U(\tau, s) \vc \xi
$$

1. It follows from points 1) and 3):

$$
    U(\tau, t) U(t, \tau) = U(\tau, \tau) = \tn I
$$



# Car acceleration - variation of constants
Consider car acceleration just for the velocity:

$$
  mv'(t) = Pt - c v(t), \qquad v(0)=0
$$

The term $Pt$ represents a linearly increasing engine force (not very realistic),
the term $-cv$ represents a friction force.

This ODE is again of the first order but is **linear** as only linear terms contain the unknown function $v$ and its derivatives.
If there is no term dependent only on $t$, we call the equation homogeneous. In our case, there is the absolute term $Pt$, so it is an inhomogeneous equation.
The absolute term may be nonlinear, or the friction coefficient $c$ could be a nonlinear function of the time, and still, the equation is linear with respect to $v$.

We usually write linear equations in a form where the terms containing $v$ are on the left and the absolute term is on the right:

$$
  mv'(t) + c v(t) = Pt 
$$

The first step is dropping of the absolute term and solution of the resulting homogeneous equation $mv'= - cv$. Using the separation of variables,
one can find the generic solution $ v(t)=Ce^{-\alpha t}$, $\alpha = c/m$. Then, we look for the solution of the original equation in the form:

$$
   v(t) = C(t) e^{-\alpha t}
$$

where we have performed *variation of the integration constant* $C$. To find the function $C(t)$, we substitute the incomplete solution into
original equation:

$$
  m C'(t) e^{-\alpha t} - m \alpha C(t) e^{-\alpha t} = Pt - m \alpha C(t) e^{-\alpha t}.
$$

We can cancel the last terms on both sides (this always happens due to a particular form of the solution), then we express $C'$ and integrate by parts:

$$
  C(t) = \frac{P}{m} \int te^{\alpha t} \d t = \frac{P}{m\alpha^2}\left(\alpha t - 1\right)e^{\alpha t} + C_0.
$$

The whole solution is then:

$$
  v(t) = C(t) e^{-\alpha t} = C_0 e^{-\alpha t} + \frac{P}{m\alpha^2} (\alpha t - 1)
$$

where $C_0$ must be determined from the initial condition. Note that the initial condition affects only the first term of the solution that corresponds to the solution
homogeneous equation, while the second term, called the **particular solution** does not depend on the initial condition, but depends on the absolute term.

## Inhomogeneous Equations

Solving inhomogeneous systems of first-order linear ODEs can be done similarly to the variation of constans 
for scalar ODE. Let us briefly recall the procedure for a scalar equation:

$$
  %\label{eq:simple_nehom}
  \dot{x} = a(t)x + b(t), \qquad x(\tau) = \xi.
$$

For the equation, we first determine the solution $x_h(t)$ of the homogeneous equation. We get a one-dimensional vector solution space

$$
   x(t) = x_h(t) \xi.
$$

Next, we look for *particular solution* using the variation of constants. 
We consider the solution in the form $x_p(t) =  x_h(t)c(t)$ and we plug it into the equation:

$$
   \dot{x}_p (t)  - a(t) x_p = b(t).
$$

Then we compute:

$$
   \dot{x}_p(t)  - a(t) x_p(t) = 
    x_h(t)\dot{c}(t) +  \dot{x}_h(t)c(t) - a(t) x_h(t) c(t) =
    x_h(t)\dot{c}(t) + \underbrace{\big(\dot{x}_h(t) - a(t) x_h(t)\big)}_{=0}c(t)\ = b(t)
$$

and so

$$
   x_p(t) = x_h(t) c(t) = x_h(t) \int_\tau^t b(s) x_h^{-1} (s) \d s.
$$

Thanks to this, the entire solution of the inhomogeneous equation is  the sum of the homogeneous and the particulate solution:

$$
   x(t) = x_h(t) \xi + x_p(t).
$$

Now we shall apply the same procedure to an inhomogeneous system of linear ODEs. 
The overall solution has the form:

$$
   \vc x(t; \tau, \vc \xi) = \tn U(t, \tau) \xi + \vc x_p(t; \tau, \vc 0)
$$

where $\tn U(t,\tau)$ is the standard fundamental matrix for the corresponding homogeneous equation and $\vc x_p$ is the particular solution of the inhomogeneous system for the zero initial vector. Following the same procedure as for the scalar case writing $\tn U(t, \tau)$ in place of $x_h(t)$, we get the formula:


$$
   \vc x_p(t; \tau, \vc 0) = \tn U(t, \tau) \int_\tau^t \big( \tn U(s,\tau)\big)^{-1} \vc b(s) \d s =
   \int_\tau^t \tn U(t, s) \vc b(s).
$$

For the second equality, just put everything under the integral and use the properties of the standard fundamental matrix:

$$
   \tn U(t, \tau) \big(\tn U(s,\tau) \big)^{-1} = \tn U(t,\tau) \tn U(\tau,s) = \tn U(t,s).
$$

These preliminary considerations allow us to state the following:


**Theorem**

Let the matrix function ${\tn A}$ and the vector function $\vc b$ be continuous on
some open interval $I$ and let $\tau \in I$ be arbitrary. Then the initial value problem?

$$
    \dot{\vc x} = {\tn A}(t)\vc x+\vc b(t),\qquad \vc x(\tau)=\vc\xi
$$

for $\xi \in \Real^n$ has exactly one solution that is on the entire interval
$I$ in the form:

$$
    \vc x(t;\tau,\vc\xi) = {\tn U}(t,\tau)\vc\xi+\int_\tau^t{\tn U}(t,s) \vc b( s)\d s,\qquad t\in I.
$$

**Proof:**
The theorem is easiest to prove by direct calculation. For the left-hand side of the equation, we have:

\begin{align*}
   \dot{\vc x}(t) &= \tn A(t) \tn U(t,\tau) \vc \xi +\tn U(t,t) \vc b(t) 
                     + \int_\tau^t \prtl_t{\tn U}(t,s) \vc b(s) \d s\\
                  &= \tn A(t) \tn U(t,\tau) \vc \xi + \vc b(t) + \int_\tau^t \tn A(t)\tn U(t,s) \vc b(s) \d s\\
                  &= \tn A(t) x(t) + b(t),
\end{align*}

where at the first line we have used the formula:

$$
   \frac{\d}{\d t} \int_\tau^t f(t,s) \d s = f(t,t) + \int_\tau^t \prtl_t f(t,s) \d s.
$$

It remains to verify the initial condition:

$$
   \vc x(\tau;\tau, \vc \xi) = {\tn U}(\tau, \tau) \vc \xi + \vc 0 = \vc \xi.
$$


## Systems with constant coefficients

Next, we will focus on the case of homogeneous linear systems with constants
coefficients, i.e. a linear system:
$$
    %\label{eq:const_coef}
    \dot{\vc x} = {\tn A}\vc x,
$$

where $\tn A$ is $n\times n$ matrix constant in time.

Since the constant matrix function $\tn A(t)=\tn A$ is indeed continuous for all $t\in \Real$,
the initial value problem with initial point $(\tau,\vc\xi)\in \Real\times \Real^n$ 
has the unique solution $\vc x(t;\tau,\vc\xi)$ defined for all times $t\in \Real$.

Since the matrix $\tn A$ does not depend on time, the solution 
(of the homogenoeus system)
will look the same for any initial time $\tau$. In particular:
$$
   \vc x(t;\tau, \vc \xi) = \vc x(t - \tau; 0, \xi),
$$

Consequently the standard fundamental matrix can be written in a simpler form:

**Proposition**

For the standard fundamental matrix $\tn U(t,\tau)$ of the ODE system with constant coefficients it holds:

$$
   {\tn U}(t,\tau)={\tn U}(t-\tau,0),\qquad \text{for all }t,\tau\in \Real.
$$

Next, we will show how to find the SFM for equations with constant coefficients.

**Proposition**
Let $\vc u\ne \vc 0$ be the eigenvector corresponding to the eigenvalue $\lambda$ of the matrix $\tn A$, i.e. it holds

$$
  \tn A \vc u = \lambda \vc u.
$$

Then the vector function

$$
\vc v(t)=\vc u e^{\lambda t},\qquad t\in \Real,
$$

is a solution to the linear system.

**Proof:**
Just plug-in $\vc v(t)$ into the equaton, we verify:

$$
  \dot{\vc v}(t) = \lambda \vc u e^{\lambda t} = \tn A \vc u e^{\lambda t} = \tn A \vc v(t).
$$


**Conclusion** If the matrix $\tn A$ has $n$ proper eigenvectors. Then the $n$ eigenvalues with $n$ corresponding eigne vectors form provide $n$ independent solutions that form the fundamental system:
$$
    \vc v_i(t) = \vc u_i e^{\lambda_i t},\quad i=1,\dots, n.
$$
This simple case happens if the system matrix $\tn A$ is diagonalizable as is explained in the follwing **algebraic comment**. If this is not the case the construction is more difficult and from the practical point of view, we suggest usage of the Putzer's method, see below.






### Algebraic comment
The eigenvalues $\lambda_i$, $i=1,\dots, m$ of the real matrix $\tn A$ are the roots of the characteristic polynomial $p(\lambda) = \det(\tn A-\lambda I)$.
This polynomial with real coefficients generally has $n$ complex roots counting each root $\lambda_i$ in its multiplicity $r_i$. To every complex root $a+ib$ there is a copmlex associated root $a-ib$.

For every invertible matrix $\tn A$ there exists an invertible matrix $\tn P$ such that

\begin{equation}
   %\label{eq:Jdecay}
   \tn A = \tn P \tn J \tn P^{-1},
\end{equation}

where $\tn J$ is the so-called [Jordan normal form](https://en.wikipedia.org/wiki/Jordan_normal_form). 
Jordan matrix $\tn J$ has the eigennumbers of the matrix $\tn A$ on the diagonal with their respective multiplicities. Moreover for some matrices it may have ones in some places of the first super-diagonal, and zeroes elsewhere. 

### Diagonalizable case
If $\tn J$ is a diagonal matrix, we say that $\tn A$ is diagonalizable. In that case, for each eigenvalue $\lambda_i$ of multiplicity $r_i$ there are $r_i$ linearly independent eigenvectors forming a subspace of dimension $r_i$. In total, we get $n$ linearly independent eigenvectors
$\vc u_j$ belonging to the eigenvalues $\lambda_j$, $j=1,\dots, n$. So the eignevalues form a basis of the vector space and these vectors satisfy the equations:

$$
   \tn A \vc u_j = \lambda_j \vc u, \qquad \text{ for all } j=1,\dots, n.
$$

For the matrix $\tn P$ in the Jordan decomposition \eqref{eq:Jdecay} one can use any matrix that has $n$ linearly independent eigenvectors of the matrix $\tn A$ in its columns, then we have

$$
    \big( \tn A \tn P \big)_{\cdot, j} = \tn A \vc u_j = \lambda_j u_j = \tn P_{\cdot, j} = \big(\tn P \tn J\big)_{\cdot, j }.
$$
We denote $\tn P_{\cdot, j}$ the $j$-th column of the matrix $\tn P$.

Moreover, if the matrix $\tn A$ is symmetric, then the eigenvectors are orthogonal and $P$ is an orthonormal matrix, i.e. $P^{-1} = P^T$.

### Example:
We have the Jordan decopmosition $\tn A = \tn P \tn J \tn P^{-1}$:

\begin{equation}
\tn A = \begin{pmatrix}
      2 & -2 \\
      0 & 2
    \end{pmatrix}
= \begin{pmatrix}
      1 & 1 \\
      0 & 1
    \end{pmatrix}
    \begin{pmatrix}
      2 & 0 \\
      0 & 2
    \end{pmatrix}
    \begin{pmatrix}
      1 & -1 \\
      0 & 1
    \end{pmatrix},
\end{equation}

with
$$
\tn P = \begin{pmatrix}
  1 & 1 \\
  0 & 1
\end{pmatrix},
\quad
\tn J = \begin{pmatrix}
  2 & 0 \\
  0 & 2
\end{pmatrix},
\quad
\tn P^{-1} = \begin{pmatrix}
  1 & -1 \\
  0 & 1
\end{pmatrix}.
$$

The Jordan block $J$ is a diagonal matrix, so the matrix $\tn A$ is diagonalizable.
However, $\tn A$ is not symmetric so $\tn P$ is not orthonormal.



### Defective case
If the Jordan form has any ones in the first subdiagonal, we say that the matrix $\tn A$ is *defective*.

### Example:
We have the Jordan decopmosition $\tn A = \tn P \tn J \tn P^{-1}$:

\begin{equation}
\tn A = \begin{pmatrix}
  2 & 3 \\
  0 & 2
\end{pmatrix} =
\begin{pmatrix}
  1 & 1 \\
  0 & 1
\end{pmatrix}
\begin{pmatrix}
  2 & 1 \\
  0 & 2
\end{pmatrix}
\begin{pmatrix}
  1 & -1 \\
  0 & 1
\end{pmatrix},
\end{equation}

with
$$
\tn P = \begin{pmatrix}
  1 & 1 \\
  0 & 1
\end{pmatrix},
\quad
\tn J = \begin{pmatrix}
  2 & 1 \\
  0 & 2
\end{pmatrix},
\quad
\tn P^{-1} = \begin{pmatrix}
  1 & -1 \\
  0 & 1
\end{pmatrix}.
$$

The Jordan block $J$ has $1$ in de first super-diagonal, so the matrix $A$ is defective.



## Putzer method
While the method of solving linear ODE systems using eigenvalues and eigenvectors is pretty easy to get the hang of, stretching it to cover the tricky cases, like those pesky repeated eigenvalues, isn't so straightforward. Practically speaking, though, this isn't a big deal, as these days you're not expected to wrestle with ODEs (even the linear ones) on paper. Instead, you'd typically use a computer for either a symbolic solution or a numerical method. But if your curiosity is tickled about an efficient and thorough algorithm for cracking linear ODE systems, the Putzer method might just be the brain candy you're looking for.

**Theorem**
Let $\lambda_1, \lambda_2,\dots,\lambda_n$ be all eigenvalues ​​of the matrix
$\tn A$ written in any order, where each eigenvalue is
written in sequence as many times as its multiplicity.

Let us construct a sequence of $n$ matrices:
\begin{align*}
  \tn P_0 &= \tn E$, \quad\text{$\tn E$ denotes the identity matrix},\\
  \tn P_1 &= ({\tn A} - \lambda_1{\tn E}){\tn P}_0 = {\tn A}- \lambda_1{\tn E}$,\\
  \dots &\\
  \tn P_j &= ({\tn A} - \lambda_j{\tn E}){\tn P}_{j-1} 
           = ({\tn A}- \lambda_n{\tn E}) ({\tn A}- \lambda_{j-1}{\tn E})\cdots({\tn A}- \lambda_1{\tn E})$,\\
  \dots&\\
  \tn P_{n-1} &= ({\tn A}- \lambda_{n-1}{\tn E}){\tn P}_{n-2} 
          = ({\tn A }- \lambda_{n-1}{\tn E})({\tn A}- \lambda_{n-2}{\tn E})\cdots({\tn A}- \lambda_1{\tn E })$,
\end{align*}
and a sequence of $n$ functions $q_j(t)$ that are solutions of $n$ Cauchy problems

$$
\begin{array}{rcll}
\dot q_1&=&\lambda_1q_1,&q_1(0)=1,\\
\dot q_2&=&\lambda_2q_2+q_1,&q_2(0)=0,\\
\dots\\
\dot q_j&=&\lambda_jq_j+q_{j-1},&q_j(0)=0,\\
\dots\\
\dot q_n&=&\lambda_nq_n+q_{n-1},&q_n(0)=0.\\
\end{array}
$$

Then the matrix

$$
{\tn U}(t) = q_1(t){\tn P}_0+q_2(t){\tn P}_1+\dots+q_n(t){\tn P}_{n-1}, ~ ~~~t\in R
$$

is the standard fundamental matrix of the Cauchy problem

$$
\dot{\vc x} = {\tn A}\vc x, \vc x(0)=\vc\xi.
$$
