```python
import numpy as np
import math
import matplotlib
import matplotlib.pyplot as plt

l = -10

def euler(y, h):
    return y *(1 + h * l)

def exact(t, t0, y0):
    return math.exp(l * (t - t0)) * y0
    
def solve(h, N, method):
    y0 = 1.0
    solution = [[0, 0, y0, 0, 0, y0]]
    y = y0
    for i in range(1, N):
        t = h * i
        yy = method(y, h)
        d = exact(t, h * (i-1), y) - yy
        e = exact(t, 0, y0) 
        g = e - yy
        solution.append([i,t,yy,d,g, e])
        y =yy
    return solution
    
def solve_to_time(h, T, method):
    return np.array(solve(h, int(T/h), method)).T

font = {'size'   : 14}

matplotlib.rc('font', **font)
fig, ax = plt.subplots(1, 2, figsize=(16,6))
for h in reversed([0.01, 0.05, 0.1, 0.2, 0.5]):  
        s = solve_to_time(h, 2.0, method=euler)
        ax[0].plot(s[1], s[2], label=f"$h = {h}$")
        ax[1].plot(s[1], s[2], label=f"$h = {h}$")

ax[0].plot(s[1], s[5], label="exact")
ax[0].set_ylim((0, 1.0))
ax[0].set_xlabel("$t$")
ax[1].set_xlabel("$t$")
ax[0].set_ylabel("$x(t)$")

ax[0].legend()
plt.show()
```

The exact solution converges quickly to zero. For small values of $h <0.1$ the numerical solution follows a similar path, while for larger time steps the numerical solution shows exponentially increasing oscillations.
The demonstrated phenomenon is called *numerical instability* of the method. 

Why is that? Let us  the choice $\lambda=-10$ and $h=0.5$, the numerical solution is given by:
$$
  y_i = (1 + h\lambda) y_{i-1} = -4 y_{i-1}= (-4)^i
$$
Therefore, the numerical solution oscillates with an exponentially increasing error.
This motivates study some stability conditions.

## Stability

Let us now introduce concept of classical stability of a numerical method:

**Definition**

A one-step numerical method is **stable** if there exist positive constants 
$\hat h$, $k$ such that for any two numerical solutions, $\vc y_i$, $\vc z_i$ of the problem with 
initial conditions $\vc y_0$ and  $\vc z_0 = \vc y_0 + \vc \delta_0$, the distance of the solutions
remains bounded as: 
$$
\max_{i} \|\vc y_i - \vc z_i\| \le k \delta,\qquad \text{for all }h\in(0,\hat h),
$$
whenever $\|\vc \delta_0\| < \delta$.

An important theoretical result is the Lax theorem, which roughly says:
$$
\text{stable} + \text{consistent} \Longrightarrow  \text{convergent}
$$

For the Euler method, we can estimate the difference of solutions as follows

\begin{align*}
  \|\vc\delta_i\| &= \vc y_i - \vc z_i \\ 
                  &\le \|\vc y_{i-1} - \vc z_{i-1}\| + h\|f(t_{i-1}, y_{i-1}) - f(t_{i-1}, z_{i-1}) \| \\
                  &\le \|\vc\delta_{i-1}\| (1 + hL)
\end{align*}

where we have used the Lipschitz condition for $f$ in the last inequality.

Iterating the inequality up to the initial condition, we get:

$$
  \|\vc\delta_i\| \le (1+hL)^i\|\vc\delta_0\| 
                  \le e^{hiL}\|\vc\delta_0\| 
                  \le k\|\vc\delta_0\| 
$$

where $k = e^{LT}$ and $\|\vc\delta_0\| \le \delta$.

Therefore, explicit Euler is a stable method !!

So why doesn't it work in the last example? 
The definition of stability is problematic for two reasons:

1. It describes the deviation of two solutions only for sufficiently small time steps $h \le \hat h$.
   As $\hat h$ only exists and could be arbitrarily small this does not prevent observed oscilations for large time steps.
   
1. It allows arbitrary (even exponential) growth of the solution deviation. So does not prevent exponential growth of error.



## Absolute stability
**Definition.**

The method is called **absolutely stable** for a time step $h$, given ODE and time interval $[0, T]$,
if changing the initial condition $y_0$ by $\delta$ causes the numerical solution 
to change on the whole time interval by at most $\delta$:

$$
    \|\vc y(t) - \vc z(t)\| \le \|\vc y_0 - \vc z_0\|,\quad \text{for all }t\in [0, T]
$$


Absolute stability is usually investigated on a scalar linear equation with a complex parameter $\lambda\in \mathbb C$:

\begin{equation}
   y'(t) = \lambda y(t),\qquad y(0) =y_0.
\end{equation}

Let us investigate absolute stability for the explicit Euler method:
$$
    y_i = y_{i-1} + h \lambda y_{i-1} = (1+h\lambda)y_{i-1}
$$
then for the difference of solutions, we have:
$$
    \delta_i = y_i - z_i = (1+h\lambda)^i |\delta_0|.
$$

To have a bounded differential for any time $t = ih\le T$, we need $|1+h\lambda|\le 1$. 
The product $h\lambda$ is a complex number that must be in the unit circle centered at $-1+0i$.

</p>

**Definition.**
The set $Z\subset \mathbb C$ for which $h\lambda\in \mathbb Z$ implies the absolute stability of the method
 is called the **region of absolute stability** of the method.


In the last example, the method should be absolutely stable for $h\lambda > -2$, which is equivalent to $h < 0.2$. 
That is what we have observed.




# Other motivation example

In the previous section, we saw that for some types of equations (with large eigenvalues), the explicit Euler method requires short time steps, leading to high computational demands and possibly large round-off error. A wise idea is to adapt the time step to error or some estimate of the second derivative of the solution. Thus, using a short timestep for time intervals where the solution changes rapidly and long time steps where it changes slowly.
This certainly helps in control of the error, but can not resolve the stability issue as we will demnostrate on the following problem:

$$  
  y' = -\alpha (y-t^2) + 2t,\quad t>0,\ y(0)=1.
$$

Its exact solution reads:

$$
   y(t)=e^{-\alpha t} + t^2.
$$

For large $\alpha=100$, the solution drops rapidly down almost to zero and then follows a parabolic shape.

```python
import matplotlib.pyplot as plt
import numpy as np

# Define the function y(t) = exp(-100t) + t^2
def y(t):
    return np.exp(-100*t) + t**2

# Create a range of t values from 0 to 0.1
t = np.linspace(0, 1, 400)

# Calculate y values
y_values = y(t)

# Create the plot
plt.figure(figsize=(8, 6))
plt.plot(t, y_values, label='y(t) = exp(-100t) + t^2')
plt.xlabel('t')
plt.ylabel('y(t)')
plt.title('Plot of y(t) = exp(-100t) + t^2')
plt.legend()
plt.grid(True)
plt.show()
```

For $t$ close to $0$, 
the first term on the right-hand side dominates and the solution 
quickly decreases to zero as $e^{-\alpha t}$. 
Then the second term dominates and the solution grows as $y \sim t^2$.

Using the explicit Euler method for this problem requires a time step $h < 2/\alpha$. 
Moreover, this condition must be satisfied
even on a time interval where the solution changes slowly. 

For comparison, consider the initial problem with only the second, slow part:
$$  
    z' = 2t, \quad t>0,\ z(0.5)=0.25
$$
Let us compare solution $z(t)$ with solution $y(t)$ for the initial condition $y(0.5) = 0.25$.


Using the time step $h=0.04$, we violate the stability condition $h<2/\alpha = 0.02$ and we observe developing oscillations in the solution $y(t)$ to the original equation even when starting from a practically exact solution.
The error is extremaly small: $\exp(-100 \cdot 0.5) \approx 2\cdot 10^{-22}$.


```python
import numpy as np
import math
import matplotlib
import matplotlib.pyplot as plt

alpha = 100

def rhs(t, y):
    return -alpha*(y-t**2) + 2*t

def exact(t, t0, y0):
    return y0 * math.exp(-alpha*t) + t**2

def euler_explicit(t, y, h):
    return y  + h * rhs(t, y)

def euler_implicit(t, y, h):
    return y  + h * rhs(t, y)

    
def solve(h, N, method):
    y0 = 1.0
    solution = [[0, 0, y0, 0, 0, y0]]
    y = y0
    for i in range(1, N):
        t = h * i
        yy = method(t, y, h)
        d = exact(t, h * (i-1), y) - yy
        e = exact(t, 0, y0) 
        g = e - yy
        solution.append([i,t,yy,d,g, e])
        y =yy
    return solution
    
def solve_to_time(h, T, method):
    return np.array(solve(h, int(T/h), method)).T

font = {'size'   : 14}

matplotlib.rc('font', **font)
fig, ax = plt.subplots(1, 2, figsize=(12,5))
for h in reversed([0.01, 0.05, 0.1, 0.2, 0.5]):  
        s = solve_to_time(h, 2.0, method=euler_explicit)
        ax[0].plot(s[1], s[2], label=f"$h = {h}$")
        ax[1].plot(s[1], s[2], label=f"$h = {h}$")

ax[0].plot(s[1], s[5], label="exact")
ax[0].set_ylim((0, 1.0))
ax[0].set_xlabel("$t$")
ax[1].set_xlabel("$t$")
ax[0].set_ylabel("$x(t)$")
ax[0].legend()
fig.suptitle('Instability for large time steps.')

plt.show()
```

```python
# Define the ODE functions and the Explicit Euler method
def ode1(y, t):
    return -100 * (y - t**2) + 2*t

def ode2(x, t):
    return 2*t

def explicit_euler(ode, y0, t0, t_end, step):
    times = np.arange(t0, t_end, step)
    y_values = [y0]
    for t in times[:-1]:
        y_values.append(y_values[-1] + step * ode(y_values[-1], t))
    return times, np.array(y_values)

# Initial conditions and parameters
y0 = 0.25
x0 = 0.25
t0 = 0.5
t_end = 1
step = 0.04

# Solve the ODEs using the Explicit Euler method
t_values, y_euler = explicit_euler(ode1, y0, t0, t_end, step)
_, x_euler = explicit_euler(ode2, x0, t0, t_end, step)

# Plotting the solutions
plt.figure(figsize=(10, 6))
plt.plot(t_values, y_euler, label="Solution of y' = -100(y-t^2)+2t", color='blue')
plt.plot(t_values, x_euler, label="Solution of x' = 2t", color='red')
plt.xlabel('t')
plt.ylabel('y(t) and x(t)')
plt.ylim((0,2))
plt.title("Comparison of Solutions using Explicit Euler Method")
plt.legend()
plt.grid(True)
plt.show()
```

## Stiff problems
A problem is called **stiff** on an interval $I$ if the time step required for absolute stability is less than the time step required to meet the desired accuracy.

A system of linear ODEs with matrix $\tn A$ is stiff if the ratio (also called **condition number**) between the largest and the smallest eigenvalue is large:
$$
\kappa(\tn A ) = \frac{\lambda_{max}}{\lambda_{min}} \gg 1
$$
How large? No strict threshold, for values above $10^2$ one should at least consider usage of an implicit method.

For a general system of equations with right-hand side $\vc f(t, \vc y)$, we can localy approximate as:
$$
\vc f(t, \vc y_0 + \vc d) = \vc f(t, \vc y_0) + \prtl_{\vc y} \vc f(t, \vc y_0) \vc d + O(|\vc d|^2)
$$
where $\prtl_{\vc y} \vc f$ is the matrix of the first derivatives called **Jacobi matrix**. Then condition number of this matrix characterize local stiffness of the equation.

**Example**
Equation
$$
    \dot{\vc y}(t) = \tn A\vc y(t),\quad 
    \tn A= \begin{pmatrix}
                0 & 1\\
                -100 & -101
           \end{pmatrix}
$$
has Jacobi matrix $\prtl_{\vc y} \vc f(t,\vc y) = \tn A$ and its eigenvalues are $-1$ and $-100$. The corresponding initial problem is rather stiff.

The restriction on the time step is common to all kinds of explicit methods.  

# Implicit Euler method
To solve stiff problems efficiently, we need a method with a large region of absolute stability so that
absolute stability criteria would not restrict the time step. As we will see later, this condition is not satisfied by any explicit method and an implicit method must be used. 

Let us consider again the general initial problem:
\begin{align*}    
   \vc y'(t) &= \vc f(t,\vc y(t)), \quad \text{for }t>0,\\
   \vc y(0)&=\vc y_0.
\end{align*}

In contrast to the explicit method, we now use the Taylor expansion at time $t_{n}$ evaluated at time 
$t_{n-1} = t_n - h$:

$$
  \vc y(t_n - h) = \vc y(t_n) - h \vc y'(t_n) + O(h^2) = \vc y_n - h \vc f(t_n, \vc y_n) + O(h^2).
$$

From here we get the equation of the implicit Euler method:

$$
    \vc y_n = \vc y_{n-1} + h\vc f(t_n, \vc y_n).
$$

The method is called *implicit* because in order to obtain the solution $y_n$ in the next time $t_n$ from the solution $y_{n-1}$, it is necessary to solve a system of nonlinear algebraic equations.

As the solution of such a nonlinear system is much more complicated than a simple application of explicit formulas, we should first prove that it would help us with absolute stability and enable us to solve stiff problems.


Applying the implicit Euler method to the linear equation we obtain:
$$
    y_n = y_{n-1} + h\lambda y_n.
$$
This is still a linear algebraic equation with the solution:
$$
    y_n = y_{n-1}\frac{1}{1-h\lambda}.
$$
The stability condition is therefore 
$$
    \frac{1}{|1-h\lambda|}\le 1,
$$
or equivalently
$$
    |1-h\lambda|\ge 1.
$$

Thus, the region of absolute stability for the implicit Euler method is the set outside the unit circle centered at $1+0i$. It looks like this:
<p align="center">
    <img src="as_implicit_euler.png" 
    alt="region of absolute stability - implicit Euler" 
    style="width:400px;"/>
</p>

We see that the method is absolutely stable for any time step as long as the equation itself is stable, i.e. $\lambda < 0$. Therefore we call such a method **unconditionally stable**. We see that the method is absolutely stable even on the positive part of the complex plane, which in fact leads to larger errors for unstable equations, like $\lambda >0$.


## Systems of nonlinear algebraic equations

We have shown the implicit Euler method to be unconditionally stable, which allows its application to stiff problems. The implicit methods, however, require the solution of a non-linear system of equations (in the general case) for every time step. To compute approximative solution $\vc Y = \vc y_n$ in time $t_n$, we have to solve the equation:  
$$
    \vc Y = \vc y_{n-1} + h \vc f(t, \vc Y)
$$
which we can also rewrite into form $\vc F(\vc Y) = 0$ as follows:
$$
    \vc F(\vc Y) = \vc Y -\vc y_{n-1} - h \vc f(t, \vc Y) 
$$


The simplest approach is the method of simple iterations:

**Simple iterations algorithm**
1. $\vc Y^0 := \vc y_{n-1}$
1. $\nu := 0$
1. $e := \infty$
1. if $e < \epsilon$ stop
1. $\vc Y^{\nu+1} := \vc y_{n-1} + h \vc f(t_n, \vc Y)$
1. $e:=\|\vc Y^{\nu+1} - \vc Y^{\nu}\|$
1. $\nu := \nu+1$
1. go to 4. 


We show that the method converges if $|hL|\le 1$, where $L$ is the Lipschitz constant of the function $\vc f$ with respect to $\vc y$. The representation of $F$ realizing 
one iteration:

$$
\vc y_n^{\nu+1}=\vc{F}(\vc{y}_n^{\nu}) = \vc y_{n-1} + h \vc f(t_n, \vc y_n^{\nu})
$$

is called contractive if 

$$
    \|\vc F(\vc x) - \vc F(\vc y)\| \le \alpha\|\vc x - \vc y\|
$$

for some $\alpha<1$. 

By Banach's fixed point theorem, the method of simple iterations will converge if $F$ is a contractive mapping.
In our case:
$$
    \|\vc F(\vc x) - \vc F(\vc y)\| = \|\vc y_{n-1} + h \vc f(t_n, \vc x) - \vc y_{n-1} - h \vc f(t_n, \vc y)\|
    \le h\|\vc f(t_n, \vc x) - \vc f(t_n, \vc y)\| \le hL\|\vc x - \vc y\|
$$
so the representation is contractive if $\alpha = hL < 1$.

We obtained here the result that the implicit Euler method using the method of simple iterations as a nonlinear solver, has the same boundary conditions on the time step
as the explicit Euler method. Thus, a better method for solving nonlinear equations is needed.



**Newton's method**

Consider the scalar nonlinear equation in form $F(Y)=0$, where $F:\Real \to \Real$ is a function with a continuous first derivative. We starts from $Y_0$, computing iterations $Y_i$. At each iteration, Newton imagines the tangent line to $F$ at point $[Y_i, F(Y_i)]$ (red in Fig.) and computes its intersection with the $y$ axis.
Following Figure depicts the process for the equation:

$$
    F(Y) = Y^3-2Y-2 = 0
$$

```python
import numpy as np
import matplotlib.pyplot as plt

# Define the nonlinear function and its derivative
def F(x):
    return x**3 - 2*x - 2

def F_prime(x):
    return 3*x**2 - 2

def plot_newton_iteration(ax, i, x0, n_iter, x_range, y_range, title):
    """
    Plots Newton's method iterations.
    
    Parameters:
    - ax: Matplotlib axis to plot on.
    - i: Iteration index for labeling.
    - x_i: Initial guess for the root.
    - n_iter: Number of iterations to perform.
    - x_range: Tuple for the x-axis range (min, max).
    - y_range: Tuple for the y-axis range (min, max).
    - title: Title of the plot.
    """
    # Generate x values for plotting the function
    x = np.linspace(*x_range, 400)
    y = F(x)
    
    # Plot the function
    ax.plot(x, y, label='F(Y)')

    x_i = x0
    # Perform Newton's method
    for iteration in range(i + n_iter):
        # Compute F(x_i) and F'(x_i)
        y_i = F(x_i)
        y_prime_i = F_prime(x_i)

        if i <= iteration:  
            # Plot tangent line at (x_i, F(x_i)) up to the x-axis
            tangent_intersect_x = x_i - y_i / y_prime_i
            tangent_x = np.linspace(x_i, tangent_intersect_x, 100)
            tangent_y = y_i + y_prime_i * (tangent_x - x_i)
            ax.plot(tangent_x, tangent_y, 'r--')
            
            # Plot vertical line connecting (x_i, 0) to (x_i, F(x_i))
            ax.plot([x_i, x_i], [0, y_i], 'b:')
        
            # Emphasize and label the iteration point
            ax.plot(x_i, y_i, 'ko', markersize=8)
            ax.text(x_i, y_i, f'$Y_{{{iteration}}}$', fontsize=12, verticalalignment='bottom', horizontalalignment='right')
        
        # Perform Newton-Raphson iteration
        x_i = x_i - y_i / y_prime_i
    
    # Add labels, title, and styling
    ax.axhline(0, color='black', linewidth=0.5)
    ax.axvline(0, color='black', linewidth=0.5)
    ax.set_xlim(*x_range)
    ax.set_ylim(*y_range)
    ax.set_xlabel('Y')
    ax.set_ylabel('F(Y)')
    ax.set_title(title)
    ax.legend()
    ax.grid(True)

# Initialize parameters
x0 = 1.2
large_scale_x_range = (1.0, 2.5)
large_scale_y_range = (-5, 8)
zoom_x_range = (1.7, 2)
zoom_y_range = (-0.1, 1.5)

# Create the figure and axes for side-by-side plots
fig, ax = plt.subplots(1, 2, figsize=(16, 6))

# Plot large-scale iterations
plot_newton_iteration(ax[0], i=0, x0=x0, n_iter=3, x_range=large_scale_x_range, 
                      y_range=large_scale_y_range, title="Newton's Method (Large Scale)")

# Plot zoomed-in iterations
plot_newton_iteration(ax[1], i=2, x0=x0, n_iter=3, x_range=zoom_x_range, 
                      y_range=zoom_y_range, title="Newton's Method (Zoomed In)")

# Adjust layout and show plots
plt.tight_layout()
plt.show()
```


We denote the initial approximation as $Y_0\in \Real$.
Further iterations of the method are computed using the formula:
$$
    Y_{\nu+1} = Y_{\nu} - \frac{F(Y_\nu)}{F'(Y_\nu)}.
$$
If the initial approximation is close enough to the solution $Y$, then the method converges very fast. However, it can fail to converge if you hit a point with a derivative close to zero.




## Newton's method for non-linear systems
To solve ODR systems, we also need to solve a system of nonlinear equations of the form:
$$
    \vc F(\vc y_n) = \vc y_n - \vc y_{n-1} -h\vc f(t_n, \vc y_n) = 0
$$

Now the derivative of $\vc F$ is a matrix:
$$
    \tn A^\nu := \partial_{\vc Y} \vc F(\vc Y^\nu) = \tn I - h \partial_{\vc y}\vc f(t_n, \vc Y^\nu)
$$
therefore, we have to rephrase the calculation of the next step as:
$$
\tn A^\nu (\vc Y^{\nu} -\vc Y^{\nu + 1}) = \vc F(\vc Y^{\nu}). 
$$

Denoting the step difference:
$$
-\vc d^{\nu+1} = \vc Y^{\nu+1} - \vc Y^{\nu}    
$$
we can formulate Newton's method for the system of equations as follows:


**Newton's method for systems**

1. $\vc y^0 := \vc y_{n-1}$
2. $\nu := 0$
3. $\vc F^{0} := \vc F(\vc Y^0) = \vc Y^0 - \vc y_{n-1} -h\vc f(t_n, \vc Y^0)$
4. if $|\vc F^{n}| < \epsilon$ stop
5. $\tn A^\nu := \partial_{\vc y} \vc F(\vc Y^\nu) = \tn I - h \partial_{\vc y}\vc f(t_n, \vc Y^\nu)$
6. solve the system: $\tn A^\nu \vc d^{\nu} = \vc F^\nu$
7. $\vc Y^{\nu+1} := \vc Y^{\nu} - \lambda \vc d^{\nu}$
8. $\vc F^{\nu+1} := \vc F(\vc Y^{\nu+1}) = \vc Y^{\nu+1} - \vc y_{n-1} -h\vc f(t_n, \vc Y^{\nu+1})$\;
9. $\nu := \nu+1$
10. go to 4.

**Additional notes**:

- The parameter $\lambda$ is equal to $1$ in the basic version of the algorithm. More advanced versions of the algorithm choose $\lambda < 1$ in order to proceed with a shortened step $\vc d^\nu$. This prevents instability of the method when $|\tn A^\nu|$ is close to zero, i.e. matrix is nearly non-invertible.

- The invertibility of $\tn A^\nu$ is characterized by its condition number. Notice that the stiffness is characterized by the condition number of the Jacobi matrix $\prtl_{\vc y} \vc f$ while here, we have condition number of $\tn A = \tn I - h \prtl_{\vc y} \vc f$.

- If the Jacobi matrix has only negative eigenvalues $-\lambda_i$, the matrix $\tn A$ has eigenvalues $1+h\lambda_i$ and its condition number is:
$$
    \kappa(\tn A) = \frac{1+h\lambda_{1}}{1+h\lambda_{n}},\qquad
    -\lambda_1 < \dots -\lambda_n < 0
$$

- We usually want $h\lambda_n \lesssim 1$ for small error, and therefore the $\kappa(\tn A)$ is usually smaller than $\kappa(\prtl_{\vc y} \vc f)$ and it tends to $1$ as $h \rightarrow 0$.

- The system of linear equations can be solved iteratively with limited accuracy. 
  
- In the context of the numerical solution of the ODE system, it is not convenient to let Newton's method use a large number of iterations, so when a suitable limit is exceeded (about 10 iterations), the time step $h$ is shortened. This will improve both the convergence of Newton's method as well as the convergence of the eventual inner linear solver.
